### plot 2d power and flux plot and save npz

```python
import os
import h5py
import numpy as np
import matplotlib.pyplot as plt

# ==========================================================
# USER SETTINGS
# ==========================================================

build_npz_from_hdf5 = True
load_npz_for_analysis = True

current_A_print = np.linspace(start=0e-3, stop=10e-3, num=81)

power_list = [0.01, 0.05, 0.1, 0.5 ,1]

batches = [
    {
        "date": "2026-07-09",
        "folder": r"C:\Users\qcrew\Desktop\Juncheng\victoria\data\2026-07-09",
        "time": ((12, 55), (13, 53)),
        "current": current_A_print,
    },
]

file_key = "RRSpecAmp_flux"

plot_data = "MAG"          # "I", "Q", or "MAG"

# Leave as None first.
# Code auto-finds the 1D frequency/cav_if axis whose length matches the last data dimension.
# If auto-find fails, set manually, e.g.
# y_key = "cav_if"
# y_key = "frequency"
# y_key = "data/frequency"
y_key = None

subtract_mean = True

x_label = "Current"
y_label = "cav_if"

plot_current_in_mA = True

output_folder = r"C:\Users\qcrew\Desktop\Juncheng\victoria\scripts\plot\data_plot_data"
os.makedirs(output_folder, exist_ok=True)

# ----------------------------
# Plot settings
# ----------------------------
plot_1d_cuts = True
save_1d_cuts = True

plot_2d_maps = True
save_2d_maps = False

plot_raw_trace_overlay = True
save_raw_trace = True

# ----------------------------
# Trace settings
# ----------------------------
find_mode = "min"        # "min" or "max"

# Generate NPZ name dynamically
st_d = batches[0]["date"].replace("-", "")
end_d = batches[-1]["date"].replace("-", "")

st_h, st_m = batches[0]["time"][0]
end_h, end_m = batches[-1]["time"][1]

save_npz_name = (
    f"{file_key}_{st_d}_{st_h:02d}{st_m:02d}"
    f"_to_{end_d}_{end_h:02d}{end_m:02d}_{plot_data}.npz"
)

npz_path = os.path.join(output_folder, save_npz_name)


# ==========================================================
# HELPER FUNCTIONS
# ==========================================================

def extract_time(filename):
    """
    Extract time from filename like:
    10-23-xx_RRSpecAmp_flux_...
    Returns (hour, minute).
    """
    try:
        base = os.path.basename(filename)
        parts = base.split("-")
        return int(parts[0]), int(parts[1])
    except Exception:
        return -1, -1


def get_filtered_files(folder, start_time, end_time, file_key):
    """
    Find all HDF5 files in folder matching file_key and time window.
    """
    if not os.path.exists(folder):
        print(f"Folder not found: {folder}")
        return []

    files = []

    for f in os.listdir(folder):
        if not f.endswith(".hdf5"):
            continue

        if file_key not in f:
            continue

        full_path = os.path.join(folder, f)
        t = extract_time(f)

        if start_time <= t <= end_time:
            files.append(full_path)

    return sorted(files, key=lambda fname: extract_time(os.path.basename(fname)))


def list_hdf5_datasets(h5_obj):
    """
    Recursively list all datasets in an HDF5 file.
    Returns list of (dataset_path, shape).
    """
    datasets = []

    def visitor(name, obj):
        if isinstance(obj, h5py.Dataset):
            datasets.append((name, obj.shape))

    h5_obj.visititems(visitor)
    return datasets


def get_dataset_by_path_or_name(f, key):
    """
    Load dataset by exact HDF5 path or by final dataset name.

    Examples:
        key = "frequency"
        key = "cav_if"
        key = "data/frequency"
    """
    if key in f:
        return np.array(f[key][()])

    all_datasets = list_hdf5_datasets(f)

    matches = []
    for path, shape in all_datasets:
        if path.split("/")[-1] == key:
            matches.append(path)

    if len(matches) == 1:
        return np.array(f[matches[0]][()])

    if len(matches) > 1:
        raise ValueError(
            f"Multiple datasets match y_key='{key}': {matches}. "
            "Use the full path, for example 'data/frequency'."
        )

    raise KeyError(f"Could not find dataset '{key}' in file.")


def find_axis_dataset(f, target_len):
    """
    Auto-find a 1D axis dataset whose length equals target_len.
    This avoids accidentally choosing the power axis of length 4.
    """
    all_datasets = list_hdf5_datasets(f)

    skip_names = {"I", "Q", "Magnitude"}

    candidates = []

    for path, shape in all_datasets:
        name = path.split("/")[-1]

        if name in skip_names:
            continue

        if len(shape) == 1 and shape[0] == target_len:
            candidates.append(path)

    if len(candidates) == 0:
        raise ValueError(
            f"Could not auto-find a 1D y-axis dataset with length {target_len}.\n"
            "Set y_key manually, for example y_key='cav_if' or y_key='frequency'."
        )

    preferred_words = [
        "freq",
        "frequency",
        "cav_if",
        "if",
        "sweep",
    ]

    for word in preferred_words:
        for c in candidates:
            if word.lower() in c.lower():
                return c, np.array(f[c][()])

    return candidates[0], np.array(f[candidates[0]][()])


def reduce_signal_array(raw, expected_power_count=None):
    """
    Convert raw HDF5 signal array into either:

        1D: (frequency,)
        2D: (power, frequency)

    Handles:
        raw shape (frequency,)              -> keep
        raw shape (power, frequency)        -> keep
        raw shape (reps, frequency)         -> average reps
        raw shape (reps, power, frequency)  -> average reps, keep power
    """
    raw = np.asarray(raw)

    if raw.ndim == 1:
        return raw

    if raw.ndim == 2:
        # Probably (power, frequency)
        if expected_power_count is not None and raw.shape[0] == expected_power_count:
            return raw

        # Otherwise probably (reps, frequency)
        return np.mean(raw, axis=0)

    if raw.ndim >= 3:
        # Probably (..., power, frequency)
        if expected_power_count is not None and raw.shape[-2] == expected_power_count:
            axes_to_average = tuple(range(raw.ndim - 2))
            return np.mean(raw, axis=axes_to_average)

        # Otherwise preserve only final frequency dimension
        axes_to_average = tuple(range(raw.ndim - 1))
        return np.mean(raw, axis=axes_to_average)

    raise ValueError(f"Unsupported raw data ndim: {raw.ndim}")


def load_one_hdf5(filename, y_key=None, subtract_mean=True):
    """
    Load one HDF5 file.

    Returns:
        selected_y_key, y_vals, I_val, Q_val, MAG_val

    Important:
        If data shape is (power, frequency), this keeps power.
        It does not average over the power axis.
    """
    with h5py.File(filename, "r") as f:

        print("\n======================================")
        print("Reading:", filename)
        print("Datasets:")

        for path, shape in list_hdf5_datasets(f):
            print(f"  {path}: {shape}")

        expected_power_count = len(power_list) if power_list is not None else None

        # --------------------------------------------------
        # Load I and Q
        # --------------------------------------------------
        if "I" not in f or "Q" not in f:
            raise KeyError("Could not find root-level 'I' and 'Q' datasets.")

        I_raw = np.array(f["I"][()])
        Q_raw = np.array(f["Q"][()])

        I_val = reduce_signal_array(I_raw, expected_power_count)
        Q_val = reduce_signal_array(Q_raw, expected_power_count)

        # --------------------------------------------------
        # Load or calculate MAG
        # --------------------------------------------------
        if "Magnitude" in f:
            MAG_raw = np.array(f["Magnitude"][()])
            mag_val = reduce_signal_array(MAG_raw, expected_power_count)
        else:
            mag_val = np.sqrt(I_val**2 + Q_val**2)

        print("Reduced I shape:", I_val.shape)
        print("Reduced Q shape:", Q_val.shape)
        print("Reduced MAG shape:", mag_val.shape)

        # --------------------------------------------------
        # Find y-axis
        # --------------------------------------------------
        target_y_len = mag_val.shape[-1]

        if y_key is not None:
            selected_y_key = y_key
            y_vals = get_dataset_by_path_or_name(f, y_key)
        else:
            selected_y_key, y_vals = find_axis_dataset(f, target_y_len)

        y_vals = np.asarray(y_vals)

        print("Selected y_key:", selected_y_key)
        print("y_vals shape:", y_vals.shape)

        if y_vals.ndim != 1:
            raise ValueError(
                f"Selected y-axis is not 1D. y_key={selected_y_key}, shape={y_vals.shape}"
            )

        if len(y_vals) != target_y_len:
            raise ValueError(
                f"Wrong y-axis length. len(y_vals)={len(y_vals)}, "
                f"but data last dimension is {target_y_len}."
            )

        # --------------------------------------------------
        # Subtract mean along frequency axis only
        # This keeps power axis.
        # --------------------------------------------------
        if subtract_mean:
            I_val = I_val - np.mean(I_val, axis=-1, keepdims=True)
            Q_val = Q_val - np.mean(Q_val, axis=-1, keepdims=True)
            mag_val = mag_val - np.mean(mag_val, axis=-1, keepdims=True)

    return selected_y_key, y_vals, I_val, Q_val, mag_val


def base_mesh_plot(x, y, z, title_suffix="", save=False, save_suffix=""):
    """
    Plot 2D mesh.

    Expected:
        x shape = (num_current,)
        y shape = (num_frequency,)
        z shape = (num_current, num_frequency)
    """
    x = np.asarray(x)
    y = np.asarray(y)
    z = np.asarray(z)

    print("plot check:")
    print("  x shape:", x.shape)
    print("  y shape:", y.shape)
    print("  z shape:", z.shape)

    if z.shape != (len(x), len(y)):
        raise ValueError(
            f"Shape mismatch: z.shape={z.shape}, "
            f"but expected ({len(x)}, {len(y)}).\n"
            "This usually means x_values or y_values is wrong."
        )

    fig, ax = plt.subplots(figsize=(9, 6))

    x_plot = x * 1e3 if plot_current_in_mA else x

    mesh = ax.pcolormesh(
        x_plot,
        y,
        z.T,
        shading="auto",
        cmap="viridis"
    )

    ax.set_xlabel(f"{x_label} ({'mA' if plot_current_in_mA else 'A'})")
    ax.set_ylabel(y_label)
    ax.set_title(f"{plot_data}: {x_label} vs {y_label} {title_suffix}")

    cbar = fig.colorbar(mesh, ax=ax)
    cbar.set_label(plot_data)

    fig.tight_layout()

    if save:
        save_name = save_npz_name.replace(".npz", f"_{save_suffix}.png")
        save_path = os.path.join(output_folder, save_name)
        fig.savefig(save_path, dpi=300)
        print(f"Saved 2D figure to: {save_path}")

    return fig, ax


def plot_1d_cuts_all_flux(x, y, z, power, save=False):
    """
    Plot all 1D cuts together.

    x: current/flux axis, shape (num_current,)
    y: frequency/cav_if axis, shape (num_frequency,)
    z: amplitude data, shape (num_current, num_frequency)

    Each line is one flux/current value.
    """
    x = np.asarray(x)
    y = np.asarray(y)
    z = np.asarray(z)

    if z.shape != (len(x), len(y)):
        raise ValueError(
            f"Shape mismatch in 1D cut plot: z.shape={z.shape}, "
            f"expected ({len(x)}, {len(y)})"
        )

    fig, ax = plt.subplots(figsize=(8, 5))

    x_color = x * 1e3 if plot_current_in_mA else x

    cmap = plt.cm.viridis
    norm = plt.Normalize(vmin=np.min(x_color), vmax=np.max(x_color))

    for i in range(len(x)):
        ax.plot(
            y,
            z[i, :],
            color=cmap(norm(x_color[i])),
            linewidth=1.2,
            alpha=0.85,
        )

    ax.set_xlabel(y_label)
    ax.set_ylabel(plot_data)
    ax.set_title(f"1D cuts: {plot_data} vs {y_label}, Power = {power}")
    ax.grid(True, alpha=0.3)

    sm = plt.cm.ScalarMappable(cmap=cmap, norm=norm)
    sm.set_array([])

    cbar = fig.colorbar(sm, ax=ax)
    cbar.set_label(f"{x_label} ({'mA' if plot_current_in_mA else 'A'})")

    fig.tight_layout()

    if save:
        power_str = str(power).replace(".", "p")
        save_name = save_npz_name.replace(
            ".npz",
            f"_P{power_str}_1D_cuts.png"
        )
        save_path = os.path.join(output_folder, save_name)
        fig.savefig(save_path, dpi=300)
        print(f"Saved 1D cuts figure to: {save_path}")

    return fig, ax


# ==========================================================
# BUILD NPZ FROM HDF5
# ==========================================================

if build_npz_from_hdf5:

    grouped_files = []
    x_values = []
    file_dates = []
    file_times = []

    for b in batches:
        files = get_filtered_files(
            folder=b["folder"],
            start_time=b["time"][0],
            end_time=b["time"][1],
            file_key=file_key,
        )

        n = min(len(files), len(b["current"]))

        grouped_files.extend(files[:n])
        x_values.extend(b["current"][:n])
        file_dates.extend([b["date"]] * n)
        file_times.extend([extract_time(f) for f in files[:n]])

    if not grouped_files:
        raise ValueError("No matching HDF5 files found.")

    print("\nMatched files:")
    for xx, ff in zip(x_values, grouped_files):
        print(f"  current = {xx*1e3:.3f} mA   file = {ff}")

    I_all = []
    Q_all = []
    MAG_all = []
    sweep_axis_all = []

    selected_y_key = None

    for i, fname in enumerate(grouped_files):
        key, y_vals, I_v, Q_v, M_v = load_one_hdf5(
            fname,
            y_key=y_key,
            subtract_mean=subtract_mean,
        )

        if selected_y_key is None:
            selected_y_key = key
            y_values = y_vals
        else:
            if len(y_vals) != len(y_values):
                raise ValueError(
                    f"File {fname} has different y-axis length: "
                    f"{len(y_vals)} vs {len(y_values)}"
                )

        sweep_axis_all.append(y_vals)
        I_all.append(I_v)
        Q_all.append(Q_v)
        MAG_all.append(M_v)

    x_values = np.asarray(x_values)
    y_values = np.asarray(y_values)

    I_all = np.asarray(I_all)
    Q_all = np.asarray(Q_all)
    MAG_all = np.asarray(MAG_all)

    print("\nFinal array shapes:")
    print("  x_values:", x_values.shape)
    print("  y_values:", y_values.shape)
    print("  I:", I_all.shape)
    print("  Q:", Q_all.shape)
    print("  MAG:", MAG_all.shape)

    np.savez_compressed(
        npz_path,

        x_values=x_values,
        current=x_values,
        current_mA=x_values * 1e3,

        y_values=y_values,
        cav_if=y_values,
        sweep_axis=np.asarray(sweep_axis_all),

        I=I_all,
        Q=Q_all,
        MAG=MAG_all,

        power_list=np.asarray(power_list),

        x_label=x_label,
        y_label=y_label,
        y_key=selected_y_key,

        filenames=np.asarray(grouped_files),
        file_dates=np.asarray(file_dates),
        file_times=np.asarray(file_times, dtype=object),

        batch_dates=np.asarray([b["date"] for b in batches]),
        batch_start_times=np.asarray([b["time"][0] for b in batches], dtype=object),
        batch_end_times=np.asarray([b["time"][1] for b in batches], dtype=object),
        batch_folders=np.asarray([b["folder"] for b in batches]),

        file_key=file_key,
        subtract_mean=subtract_mean,
        plot_data_saved=plot_data,
    )

    print(f"\nSaved main NPZ to: {npz_path}")


# ==========================================================
# LOAD NPZ AND ANALYZE
# ==========================================================

if load_npz_for_analysis:

    data = np.load(npz_path, allow_pickle=True)

    x_values = np.asarray(data["current"])

    if "cav_if" in data.keys():
        y_values = np.asarray(data["cav_if"])
    else:
        y_values = np.asarray(data["y_values"])

    data_all = np.asarray(data[plot_data])

    print("\nLoaded NPZ:")
    print("  data shape:", data_all.shape)
    print("  x_values shape:", x_values.shape)
    print("  y_values shape:", y_values.shape)

    # -------------------------------------------------------
    # Automatically handle 2D and 3D datasets
    # -------------------------------------------------------
    if data_all.ndim == 2:
        # Shape: current, frequency
        data_all = data_all[:, np.newaxis, :]
        power_values = [0]

    elif data_all.ndim == 3:
        # Shape: current, power, frequency
        if "power_list" in data.keys():
            power_values = list(data["power_list"])
        else:
            power_values = power_list

        if len(power_values) != data_all.shape[1]:
            print(
                "Warning: len(power_values) does not match data power dimension. "
                "Using index numbers instead."
            )
            power_values = list(range(data_all.shape[1]))

    else:
        raise ValueError(
            f"Unsupported data_all.ndim={data_all.ndim}. "
            "Expected 2D or 3D data."
        )

    # Final safety check
    if len(x_values) != data_all.shape[0]:
        raise ValueError(
            f"x-axis mismatch: len(x_values)={len(x_values)}, "
            f"but data current dimension={data_all.shape[0]}"
        )

    if len(y_values) != data_all.shape[-1]:
        raise ValueError(
            f"y-axis mismatch: len(y_values)={len(y_values)}, "
            f"but data frequency dimension={data_all.shape[-1]}. "
            "Your y-axis is probably the power axis, not the frequency axis."
        )

    # =======================================================
    # Loop over every power
    # =======================================================

    for p_index, power in enumerate(power_values):

        print(f"\n========== Power = {power} ==========")

        data_2d = data_all[:, p_index, :]

        print("data_2d shape:", data_2d.shape)

        # ----------------------------
        # 1D cuts: all flux/current traces together
        # ----------------------------
        if plot_1d_cuts:
            plot_1d_cuts_all_flux(
                x_values,
                y_values,
                data_2d,
                power,
                save=save_1d_cuts,
            )
            plt.show()

        # ----------------------------
        # Raw 2D plot
        # ----------------------------
        if plot_2d_maps:
            power_str = str(power).replace(".", "p")

            base_mesh_plot(
                x_values,
                y_values,
                data_2d,
                title_suffix=f"(Power = {power})",
                save=save_2d_maps,
                save_suffix=f"P{power_str}_2D_raw",
            )
            plt.show()

        # ----------------------------
        # Find raw trace
        # ----------------------------
        if find_mode == "min":
            idx = np.argmin(data_2d, axis=1)
        elif find_mode == "max":
            idx = np.argmax(data_2d, axis=1)
        else:
            raise ValueError("find_mode must be 'min' or 'max'.")

        trace_y = y_values[idx]
        trace_val = data_2d[np.arange(len(idx)), idx]

        # ----------------------------
        # Overlay raw trace
        # ----------------------------
        if plot_raw_trace_overlay:
            base_mesh_plot(
                x_values,
                y_values,
                data_2d,
                title_suffix=f"(Power = {power}, raw {find_mode})",
                save=False,
                save_suffix="",
            )

            plt.plot(
                x_values * 1e3 if plot_current_in_mA else x_values,
                trace_y,
                "r.",
                markersize=4,
                label=f"Raw {find_mode} trace"
            )

            plt.legend()
            plt.show()

        # ----------------------------
        # Save raw trace NPZ
        # ----------------------------
        if save_raw_trace:

            power_str = str(power).replace(".", "p")

            trace_save_name = save_npz_name.replace(
                ".npz",
                f"_P{power_str}_{find_mode}_raw_trace.npz"
            )

            trace_save_path = os.path.join(output_folder, trace_save_name)

            np.savez_compressed(
                trace_save_path,

                current=x_values,
                current_mA=x_values * 1e3,

                cav_if=trace_y,
                y_values=trace_y,

                value=trace_val,
                power=power,
                plot_data=plot_data,
                find_mode=find_mode,
            )

            print(f"Saved raw trace NPZ to: {trace_save_path}")

        print("\ncurrent (mA):")
        print(x_values * 1e3)

        print("\ncav_if raw trace:")
        print(trace_y)
```

```python
import os
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# ==========================================================
# USER SETTINGS
# ==========================================================

output_folder = r"C:\Users\qcrew\Desktop\Juncheng\victoria\scripts\plot\data_plot_data"

npz_name = "RRSpecAmp_flux_20260709_1023_to_20260709_1053_MAG.npz"
npz_path = os.path.join(output_folder, npz_name)

plot_data = "MAG"
target_power = 0.01

# Use "min" if resonance is a dip in MAG.
# Use "max" if resonance is a peak in MAG.
resonance_find_mode = "min"

# Optional current window for fitting
use_current_window = False
current_min_mA_for_fit = -20
current_max_mA_for_fit = 20

# If you roughly know one Phi0 period, put it here in mA.
# If None, code tries many guesses automatically.
expected_I_phi0_mA = None

# Outlier rejection after first cosine fit
reject_outliers = True
outlier_sigma = 3.0

# ==========================================================
# LOAD NPZ ONLY
# ==========================================================

data = np.load(npz_path, allow_pickle=True)

print("Loaded NPZ:")
print(npz_path)

print("\nKeys:")
print(data.files)

current = np.asarray(data["current"])      # A
current_mA = current * 1e3                 # mA

if "rrfreq" in data.files:
    rrfreq = np.asarray(data["rrfreq"])
elif "cav_if" in data.files:
    rrfreq = np.asarray(data["cav_if"])
else:
    rrfreq = np.asarray(data["y_values"])

data_all = np.asarray(data[plot_data])
power_values = np.asarray(data["power_list"])

print("\nShapes:")
print("  current:", current.shape)
print("  rrfreq:", rrfreq.shape)
print(f"  {plot_data}:", data_all.shape)
print("  power_values:", power_values)

# ==========================================================
# MAKE DATA SHAPE current, power, rrfreq
# ==========================================================

if data_all.ndim == 2:
    data_all = data_all[:, np.newaxis, :]
    power_values = np.array([0])

if data_all.ndim != 3:
    raise ValueError("Expected data shape: current, power, rrfreq")

# ==========================================================
# SELECT POWER = 0.01
# ==========================================================

p_index = np.argmin(np.abs(power_values - target_power))
selected_power = power_values[p_index]

print("\nSelected power:")
print("  target_power =", target_power)
print("  selected_power =", selected_power)
print("  p_index =", p_index)

data_power = data_all[:, p_index, :]   # shape: current, rrfreq

# ==========================================================
# EXTRACT RESONANCE TRACE
# ==========================================================

if resonance_find_mode == "min":
    rr_idx = np.argmin(data_power, axis=1)
elif resonance_find_mode == "max":
    rr_idx = np.argmax(data_power, axis=1)
else:
    raise ValueError("resonance_find_mode must be 'min' or 'max'.")

rrfreq_trace = rrfreq[rr_idx]
amp_trace = data_power[np.arange(len(current)), rr_idx]

# ==========================================================
# FIT MASK
# ==========================================================

fit_mask = np.isfinite(current) & np.isfinite(rrfreq_trace)

if use_current_window:
    fit_mask &= (
        (current_mA >= current_min_mA_for_fit)
        &
        (current_mA <= current_max_mA_for_fit)
    )

current_fit = current[fit_mask]
current_mA_fit = current_mA[fit_mask]
rrfreq_fit = rrfreq_trace[fit_mask]

if len(current_fit) < 5:
    raise ValueError("Not enough points for cosine fitting.")

# ==========================================================
# COSINE MODEL
# Higher frequency = 0 flux quanta
# ==========================================================

def cos_model(I, f_center, f_amp, I0, I_phi0):
    """
    I: current in A
    f_center: center frequency
    f_amp: positive amplitude
    I0: current where frequency is maximum = 0 flux
    I_phi0: one flux quantum current period
    """
    return f_center + f_amp * np.cos(2 * np.pi * (I - I0) / I_phi0)


def fit_cosine_many_guesses(I, y):
    """
    Robust-ish cosine fit using many initial period and phase guesses.
    """
    I = np.asarray(I)
    y = np.asarray(y)

    I_range = np.max(I) - np.min(I)
    if I_range <= 0:
        raise ValueError("Current range is zero.")

    f_center0 = np.mean(y)
    f_amp0 = 0.5 * (np.max(y) - np.min(y))

    if f_amp0 == 0:
        f_amp0 = 1e-6

    # Raw highest-frequency point is only used as an initial phase guess
    I_high_raw = I[np.argmax(y)]

    # Period guesses
    if expected_I_phi0_mA is not None:
        P0 = expected_I_phi0_mA * 1e-3
        period_guesses = np.array([
            0.5 * P0,
            0.8 * P0,
            P0,
            1.2 * P0,
            1.5 * P0,
        ])
    else:
        # Try broad range of periods
        period_guesses = np.linspace(0.5 * I_range, 4.0 * I_range, 40)

        # Also add raw high-low estimate if useful
        I_low_raw = I[np.argmin(y)]
        P_raw = 2 * abs(I_low_raw - I_high_raw)
        if P_raw > 0:
            period_guesses = np.unique(np.append(period_guesses, P_raw))

    # Phase guesses around high-frequency current
    phase_guesses = []
    for P in period_guesses:
        phase_guesses.extend([
            I_high_raw,
            I_high_raw - P,
            I_high_raw + P,
            np.mean(I),
            0.0,
        ])

    # Bounds
    P_min = 0.2 * I_range
    P_max = 8.0 * I_range

    lower = [
        np.min(y) - 2 * abs(f_amp0),
        0.0,
        np.min(I) - 4 * I_range,
        P_min,
    ]

    upper = [
        np.max(y) + 2 * abs(f_amp0),
        10 * abs(f_amp0) + 1e-9,
        np.max(I) + 4 * I_range,
        P_max,
    ]

    best_popt = None
    best_pcov = None
    best_rss = np.inf

    for P_guess in period_guesses:
        if not (P_min <= P_guess <= P_max):
            continue

        for I0_guess in phase_guesses:
            p0 = [
                f_center0,
                abs(f_amp0),
                I0_guess,
                P_guess,
            ]

            # Clip p0 into bounds
            p0 = np.maximum(p0, lower)
            p0 = np.minimum(p0, upper)

            try:
                popt, pcov = curve_fit(
                    cos_model,
                    I,
                    y,
                    p0=p0,
                    bounds=(lower, upper),
                    maxfev=50000,
                )

                residual = y - cos_model(I, *popt)
                rss = np.sum(residual**2)

                if rss < best_rss:
                    best_rss = rss
                    best_popt = popt
                    best_pcov = pcov

            except Exception:
                pass

    if best_popt is None:
        raise RuntimeError("Cosine fit failed for all initial guesses.")

    return best_popt, best_pcov, best_rss


# ==========================================================
# FIRST FIT
# ==========================================================

popt, pcov, rss = fit_cosine_many_guesses(current_fit, rrfreq_fit)

f_center, f_amp, I0_fit, I_phi0_fit = popt

# ==========================================================
# OPTIONAL OUTLIER REJECTION + REFIT
# ==========================================================

good_fit_mask = np.ones(len(current_fit), dtype=bool)

if reject_outliers:
    y_fit_first = cos_model(current_fit, *popt)
    residual = rrfreq_fit - y_fit_first

    med = np.median(residual)
    mad = np.median(np.abs(residual - med))

    if mad > 0:
        sigma_est = 1.4826 * mad
        good_fit_mask = np.abs(residual - med) < outlier_sigma * sigma_est

        if np.sum(good_fit_mask) >= 5:
            popt, pcov, rss = fit_cosine_many_guesses(
                current_fit[good_fit_mask],
                rrfreq_fit[good_fit_mask],
            )

            f_center, f_amp, I0_fit, I_phi0_fit = popt

# ==========================================================
# NORMALIZE ZERO-FLUX OFFSET
# ==========================================================
# I0_fit + n * I_phi0_fit are equivalent 0-flux points.
# Choose the equivalent 0-flux current closest to 0 mA.

n_near_zero = np.round((0.0 - I0_fit) / I_phi0_fit)
I0_near_zero = I0_fit + n_near_zero * I_phi0_fit

I0_fit_mA = I0_fit * 1e3
I0_near_zero_mA = I0_near_zero * 1e3
I_phi0_mA = I_phi0_fit * 1e3

# Flux axis using fitted cosine
flux_quanta = (current - I0_near_zero) / I_phi0_fit
flux_quanta_mod = np.mod(flux_quanta, 1.0)

flux_at_zero_current = (0.0 - I0_near_zero) / I_phi0_fit
flux_at_zero_current_mod = np.mod(flux_at_zero_current, 1.0)

# Low-frequency point is half flux away from high-frequency point
I_half_flux = I0_near_zero + 0.5 * I_phi0_fit
I_half_flux_mA = I_half_flux * 1e3

# Dense fit curve
current_dense = np.linspace(np.min(current), np.max(current), 1000)
current_dense_mA = current_dense * 1e3
fit_dense = cos_model(current_dense, *popt)
flux_dense = (current_dense - I0_near_zero) / I_phi0_fit

fit_on_points = cos_model(current, *popt)

# ==========================================================
# PRINT RESULTS
# ==========================================================

print("\n========== Cosine Fit Result ==========")
print(f"Power = {selected_power}")

print("\nFit parameters:")
print(f"  f_center = {f_center}")
print(f"  f_amp    = {f_amp}")
print(f"  I_phi0   = {I_phi0_mA:.6f} mA")

print("\n0-flux offset:")
print(f"  fitted I0 raw              = {I0_fit_mA:.6f} mA")
print(f"  equivalent I0 near 0 mA    = {I0_near_zero_mA:.6f} mA")
print(f"  current for 0.5 flux       = {I_half_flux_mA:.6f} mA")

print("\n0-current flux offset:")
print(f"  flux at 0 current       = {flux_at_zero_current:.6f} Phi0")
print(f"  flux at 0 current mod 1 = {flux_at_zero_current_mod:.6f} Phi0")

print("\nFit quality:")
print(f"  RSS = {rss}")
print(f"  used points after outlier rejection = {np.sum(good_fit_mask)} / {len(current_fit)}")

print("\ncurrent_mA:")
print(current_mA)

print("\nflux_quanta:")
print(flux_quanta)

print("\nflux_quanta_mod:")
print(flux_quanta_mod)

# ==========================================================
# PLOT 1: RESONANCE TRACE VS CURRENT WITH COS FIT
# ==========================================================

plt.figure(figsize=(8, 5))

plt.plot(
    current_mA,
    rrfreq_trace,
    "o",
    markersize=4,
    label="Extracted resonance",
)

plt.plot(
    current_dense_mA,
    fit_dense,
    "-",
    linewidth=2,
    label="Cosine fit",
)

plt.axvline(
    I0_near_zero_mA,
    color="r",
    linestyle="--",
    label="0 flux, high freq",
)

plt.axvline(
    I_half_flux_mA,
    color="b",
    linestyle="--",
    label="0.5 flux, low freq",
)

plt.axvline(
    0,
    color="k",
    linestyle=":",
    label=f"0 current = {flux_at_zero_current_mod:.3f} Phi0 mod 1",
)

plt.xlabel("Current (mA)")
plt.ylabel("rrfreq")
plt.title(f"Power = {selected_power}, cosine fit")
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================================
# PLOT 2: PCOLORMESH VS CURRENT WITH TRACE AND COS FIT
# ==========================================================

plt.figure(figsize=(9, 6))

plt.pcolormesh(
    current_mA,
    rrfreq,
    data_power.T,
    shading="auto",
    cmap="viridis",
)

plt.plot(
    current_mA,
    rrfreq_trace,
    "r.",
    markersize=5,
    label="Extracted resonance",
)

plt.plot(
    current_dense_mA,
    fit_dense,
    "w-",
    linewidth=2,
    label="Cosine fit",
)

plt.axvline(
    I0_near_zero_mA,
    color="w",
    linestyle="--",
    linewidth=1.5,
    label="0 flux",
)

plt.axvline(
    I_half_flux_mA,
    color="cyan",
    linestyle="--",
    linewidth=1.5,
    label="0.5 flux",
)

plt.axvline(
    0,
    color="k",
    linestyle=":",
    linewidth=2,
    label=f"0 current = {flux_at_zero_current_mod:.3f} Phi0",
)

plt.xlabel("Current (mA)")
plt.ylabel("rrfreq")
plt.title(f"{plot_data}, power = {selected_power}: current vs rrfreq")
plt.colorbar(label=plot_data)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================================
# PLOT 3: PCOLORMESH VS FLUX WITH TRACE AND COS FIT
# ==========================================================

sort_idx = np.argsort(flux_quanta)

flux_sorted = flux_quanta[sort_idx]
rrfreq_trace_sorted = rrfreq_trace[sort_idx]
data_power_sorted = data_power[sort_idx, :]

plt.figure(figsize=(9, 6))

plt.pcolormesh(
    flux_sorted,
    rrfreq,
    data_power_sorted.T,
    shading="auto",
    cmap="viridis",
)

plt.plot(
    flux_sorted,
    rrfreq_trace_sorted,
    "r.",
    markersize=5,
    label="Extracted resonance",
)

plt.plot(
    flux_dense,
    fit_dense,
    "w-",
    linewidth=2,
    label="Cosine fit",
)

plt.axvline(
    0,
    color="w",
    linestyle="--",
    linewidth=1.5,
    label="0 flux, high freq",
)

plt.axvline(
    0.5,
    color="cyan",
    linestyle="--",
    linewidth=1.5,
    label="0.5 flux, low freq",
)

plt.axvline(
    flux_at_zero_current,
    color="k",
    linestyle=":",
    linewidth=2,
    label="0 current",
)

plt.xlabel(r"Flux $\Phi / \Phi_0$")
plt.ylabel("rrfreq")
plt.title(f"{plot_data}, power = {selected_power}: flux vs rrfreq")
plt.colorbar(label=plot_data)
plt.legend()
plt.tight_layout()
plt.show()

# ==========================================================
# SAVE RESULT
# ==========================================================

# save_path = npz_path.replace(
#     ".npz",
#     f"_P{str(selected_power).replace('.', 'p')}_cosfit_flux_calibrated.npz"
# )

# np.savez_compressed(
#     save_path,

#     current=current,
#     current_mA=current_mA,

#     rrfreq=rrfreq,
#     rrfreq_trace=rrfreq_trace,
#     amp_trace=amp_trace,

#     selected_power=selected_power,
#     power_index=p_index,

#     fit_on_points=fit_on_points,
#     current_dense=current_dense,
#     current_dense_mA=current_dense_mA,
#     fit_dense=fit_dense,
#     flux_dense=flux_dense,

#     f_center=f_center,
#     f_amp=f_amp,

#     I0_fit=I0_fit,
#     I0_fit_mA=I0_fit_mA,

#     I0_near_zero=I0_near_zero,
#     I0_near_zero_mA=I0_near_zero_mA,

#     I_phi0=I_phi0_fit,
#     I_phi0_mA=I_phi0_mA,

#     I_half_flux=I_half_flux,
#     I_half_flux_mA=I_half_flux_mA,

#     flux_at_zero_current=flux_at_zero_current,
#     flux_at_zero_current_mod=flux_at_zero_current_mod,

#     flux_quanta=flux_quanta,
#     flux_quanta_mod=flux_quanta_mod,

#     good_fit_mask=good_fit_mask,
#     rss=rss,

#     plot_data=plot_data,
#     resonance_find_mode=resonance_find_mode,
# )

# print("\nSaved cosine-fit calibrated NPZ:")
# print(save_path)
```

```python
# ==========================================================
# FLUX <-> CURRENT CONVERSION HELPERS
# ==========================================================
# Definition:
#   flux = 0 means highest-frequency point
#   I0_near_zero is the current offset for 0 flux
#   I_phi0_fit is one flux quantum period in current

def flux_to_current_mA(phi, branch=0):
    """
    Convert flux quanta to current in mA.

    phi:
        Flux in Phi/Phi0.
        Example: phi = 0.4

    branch:
        Integer period branch.
        branch = 0 gives current near I0_near_zero.
        branch = 1 gives one Phi0 period higher.
        branch = -1 gives one Phi0 period lower.
    """
    I_A = I0_near_zero + (phi + branch) * I_phi0_fit
    return I_A * 1e3


def current_mA_to_flux(I_mA, mod=True):
    """
    Convert current in mA to flux quanta.

    mod=True:
        Return flux modulo 1, between 0 and 1.

    mod=False:
        Return signed/unwrapped flux.
    """
    I_A = I_mA * 1e-3
    phi = (I_A - I0_near_zero) / I_phi0_fit

    if mod:
        return np.mod(phi, 1.0)
    else:
        return phi


# ==========================================================
# PRINT CURRENT OFFSET FOR 0 FLUX
# ==========================================================

print("\n========== Flux-current conversion ==========")

print(f"Current offset for 0 flux:")
print(f"  I_0flux = {I0_near_zero_mA:.9f} mA")

print(f"\nOne flux quantum period:")
print(f"  I_phi0 = {I_phi0_mA:.9f} mA")

print(f"\nFlux offset at 0 current:")
print(f"  phi(I = 0 mA)       = {flux_at_zero_current:.9f} Phi0")
print(f"  phi(I = 0 mA) mod 1 = {flux_at_zero_current_mod:.9f} Phi0")


# ==========================================================
# EXAMPLE 1: SEARCH 0.4 FLUX QUANTA -> CURRENT
# ==========================================================

target_phi = -0.4

print(f"\nCurrent corresponding to phi = {target_phi} Phi0:")

for branch in [-2, -1, 0, 1, 2]:
    I_target_mA = flux_to_current_mA(target_phi, branch=branch)
    print(f"  branch {branch:+d}: I = {I_target_mA:.9f} mA")


# ==========================================================
# EXAMPLE 2: SEARCH CURRENT -> FLUX QUANTA
# ==========================================================

target_current_mA = 0.0

phi_unwrapped = current_mA_to_flux(target_current_mA, mod=False)
phi_mod = current_mA_to_flux(target_current_mA, mod=True)

print(f"\nFlux corresponding to I = {target_current_mA} mA:")
print(f"  unwrapped phi = {phi_unwrapped:.9f} Phi0")
print(f"  mod-1 phi     = {phi_mod:.9f} Phi0")
```
